# 3. Baselines
Pitch-histogram and pitch-bigram classifiers on the same five splits. No GPU is needed.

```python
from constants import *
from baselines import run_baselines, FEATURE_KINDS, CLASSIFIER_NAMES
from disk_utils import load_results
from result_tables import per_seed_table, summary_table
```

```python
# QUICK_TEST = True runs split 0 only and writes to results_quick_test/.
QUICK_TEST = False

if QUICK_TEST:
    results_dir = QUICK_TEST_RESULTS_DIR
    seeds = [0]
else:
    results_dir = RESULTS_DIR
    seeds = SEEDS
```

Feature sets: `hist53_dur` (duration-weighted pitch-class histogram, 53 bins), `hist53_cnt` (the same with note counts), `hist_abs_dur` (duration-weighted histogram of absolute pitches, 265 bins), `bigram` (TF-IDF of pitch bigrams).

Classifiers: logistic regression, SVM with RBF kernel, random forest, k-nearest neighbours. The setting of each classifier is selected on the validation subset.

```python
print(FEATURE_KINDS)
print(CLASSIFIER_NAMES)
```

```python
run_baselines(
    seeds=seeds,
    feature_kinds=FEATURE_KINDS,
    classifier_names=CLASSIFIER_NAMES,
    results_dir=results_dir
)
```
